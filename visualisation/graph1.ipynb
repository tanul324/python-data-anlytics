```python
import pandas as pandas 
import seaborn as sns  
import numpy as np 
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('titanic')
```

```python
#countplot : survival based on gender
sns.countplot(df, x='survived' ,hue='sex')
plt.title('survived vs not survived by gender')
plt.show()
```

```python
sns.barplot(df ,x ='class',y='fare', hue ='sex')
plt.title('average Fare by passanger class and gender')
plt.show()
```

```python
sns.barplot(df, x= 'class',y='age',hue='sex')
plt.title('age distribution by passenger class and gender')
plt.show()
```

```python
sns.pointplot(df,x='class',y='survived',hue='sex' ,palette='dark',markers=['o','x'], linestyles=['-','--'])
plt.title('survival rate by class and gender')
plt.show()
```
